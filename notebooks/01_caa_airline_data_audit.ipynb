```python
def get_caa_table03(year, month):
    page_url = f"https://www.caa.co.uk/data-and-analysis/uk-aviation-market/airlines/uk-airline-data/uk-airline-data-{year}/{month}-{year}/"

    response = requests.get(page_url)

    soup = BeautifulSoup(response.text, "html.parser")

    for link in soup.find_all("a"):
        if "Table 03" in link.get_text() and "CSV" in link.get_text():
            csv_path = link.get("href")

    full_csv_url = "https://www.caa.co.uk" + csv_path

    df = pd.read_csv(full_csv_url)

    return df
```

```python
months = pd.date_range(
    start="2019-01-01",
    end="2026-07-01",
    freq="MS"
)

len(months)
```

```python
for date in months[:5]:
    print(date.year, date.strftime("%B").lower())
```

```python
test_datasets = []

for date in months[:3]:
    year = date.year
    month = date.strftime("%B").lower()

    df_month = get_caa_table03(year, month)

    test_datasets.append(df_month)

    print(year, month, df_month.shape)
```

```python
test_combined = pd.concat(test_datasets, ignore_index=True)

test_combined.shape
```

```python
all_datasets = []
failed_months = []

for date in months:
    year = date.year
    month = date.strftime("%B").lower()

    try:
        df_month = get_caa_table03(year, month)
        all_datasets.append(df_month)
        print(f"Success: {year} {month}")

    except Exception as error:
        failed_months.append((year, month, str(error)))
        print(f"Failed: {year} {month} - {error}")
```

```python
print("Successful months:", len(all_datasets))
print("Failed months:", len(failed_months))
print("Failed details:", failed_months)
```

```python
caa_historical = pd.concat(all_datasets, ignore_index=True)

caa_historical.shape
```
